```python
s = input("Enter a string: ")
if s == s[::-1]:
    print("Palindrome")
else:
    print("Not Palindrome")
```

```python
s = input("Enter a string: ")
vowels = "aeiouAEIOU"
v = 0
c = 0
for ch in s:
    if ch.isalpha():
        if ch in vowels:
            v += 1
        else:
            c += 1
print("Vowels:", v)
print("Consonants:", c)
```

```python
s = input("Enter a sentence: ")
old = input("Enter word to replace: ")
new = input("Enter new word: ")
result = s.replace(old, new)
print("Updated sentence:", result)
```

```python
s = input("Enter a string: ")
for ch in s:
    print(ch, ":", s.count(ch))
```

```python
s = input("Enter a string: ")
checked = ""
for ch in s:
    if ch not in checked:
        print(ch, ":", s.count(ch))
        checked += ch
```

```python
s = input("Enter a string: ")
start = int(input("Enter start index: "))
end = int(input("Enter end index: "))
sub = s[start:end]
print("Substring:", sub)
```

```python
s1 = input("Enter first string: ")
s2 = input("Enter second string: ")
if sorted(s1) == sorted(s2):
    print("Anagrams")
else:
    print("Not Anagrams")
```
